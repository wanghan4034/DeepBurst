```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

eid = 'E116'

MARKS = ['H3K4me1', 'H3K4me3', 'H3K9me3', 'H3K27me3', 'H3K36me3', 'H3K27ac', 'H3K9ac']
yticks = ['H3K9ac', 'H3K27me3']

fig, ax = plt.subplots(2, 4, constrained_layout=True, figsize=(10/2.54, 3.5/2.54))

# Loop through the plots
for idx1, mark in enumerate(yticks):
    _idx = MARKS.index(mark)

    for idx2, target in enumerate(['mean', 'bf', 'bs', 'cv']):
        high_marks = np.load(f"extra/datasets/benchmark/Promoterformer/results/{eid}_{target}_para_highmarks.npy")
        low_marks = np.load(f"extra/datasets/benchmark/Promoterformer/results/{eid}_{target}_para_lowmarks.npy")

        # 绘制折线图
        sns.lineplot(x=range(len(high_marks[_idx])), y=high_marks[_idx], color='blue', ax=ax[idx1, idx2])
        sns.lineplot(x=range(len(low_marks[_idx])), y=low_marks[_idx], color=(177/255, 193/255, 225/255), ax=ax[idx1, idx2])

        # 显示最大差值
        diff_value = round(max(abs(high_marks[_idx] - low_marks[_idx])), 2)
        if idx2 < 3:
            ax[idx1, idx2].text(50, 2.1 if idx2 < 3 else 1.7, diff_value, fontdict={"size": 5}, color='blue') if idx1==0 else ax[idx1, idx2].text(50, 0.9 if idx2 < 3 else 1.7, diff_value, fontdict={"size": 5}, color='blue') 
        else:
            ax[idx1, idx2].text(40, 2.1, f"{round(max(abs(high_marks[_idx]-low_marks[_idx])),2)} = max|$BS^+$-$BS^-$|",fontdict={"size":5}, color='blue') if idx1==0 else ax[idx1, idx2].text(40, 0.9, f"{round(max(abs(high_marks[_idx]-low_marks[_idx])),2)} = max|$BS^+$-$BS^-$|",fontdict={"size":5}, color='blue')

        # 设定 y 轴刻度
        y_ticks_range = [i * 0.1 for i in range(0, 26, 5)] if idx1 == 0 else [i * 0.1 for i in range(0, 11, 5)]
        if idx2 == 0:
            ax[idx1, idx2].set_yticks(y_ticks_range,labels=y_ticks_range)
        else:
            ax[idx1, idx2].set_yticks(y_ticks_range,labels=[])

        # 设置标题
        x_ticks = [i for i in range(0, 81, 40)]
        if idx1 == 0:
            ax[idx1, idx2].set_title(target, fontsize=6, fontweight='regular')
            ax[idx1, idx2].set_xticks(x_ticks, labels='')
        else:
           
            labels=[int((i-40)/2) for i in range(0, 81, 40)]
            ax[idx1, idx2].set_xticks(x_ticks, labels=labels)

        # 设置 y 轴标签
        if idx2 == 0:
            ax[idx1, idx2].set_ylabel(mark, fontsize=5)


        # 设定 ticks 样式
        ax[idx1, idx2].tick_params(axis='both', labelsize=5, width=0.5)

        # 设定 y 轴限制
        ax[idx1, idx2].set_ylim(ymax=2.6 if idx1 == 0 else 1.1)

# 添加图例
lines = ax[0, 0].get_lines()
handles = [lines[0], lines[1]]
labels = ["$+$", "$-$"]
fig.legend(handles, labels, loc='upper right', bbox_to_anchor=(1.1, 1.0), title='Target', title_fontsize=6, fontsize=5, ncol=1, edgecolor='none')

# 设定总标题
fig.supxlabel("Distance to TSS (kbp)", fontsize=6)
fig.supylabel(f"Signal strength ({EIDS_TO_CELLTYPES[eid]})", fontsize=6, y=0.55, x=-0.02)

# 保存图片
plt.savefig(f'extra/datasets/benchmark/Promoterformer/figures/Figure5B.eps', format='eps', bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# Global config
# -----------------------------

eid = "E116"
save_path = f"extra/datasets/benchmark/Promoterformer/figures/v6/Figure5A.eps"

# perturbation 顺序（10 表示 1.0，0 表示 0.0）
order = list(range(10, -1, -1))
# 显示为 1.0、0.9、…、0.0；隔一个标一个
xticklabels = [f"{1 - i*0.1:.1f}" if i % 2 == 0 else "" for i in order]

# 调色
palette_box = ['#E94C72', '#036EB8']   # BF, BS
bar_low_color  = '#ECD698'
bar_high_color = '#E2AC91'

# -----------------------------
# Load once
# -----------------------------
df_mean = pd.read_csv(f"extra/datasets/benchmark/Promoterformer/results/{eid}_perturbation_predictions_mean.csv")
df_bfbs = pd.read_csv(f"extra/datasets/benchmark/Promoterformer/results/{eid}_perturbation_predictions_bs_bf.csv")

# 保障类型与顺序稳定：把 perturbation_strength 映射为有序分类
cat = pd.CategoricalDtype(categories=order, ordered=True)
for d in (df_mean, df_bfbs):
    d["perturbation_strength"] = d["perturbation_strength"].astype(int).astype(cat)

# -----------------------------
# Small helpers
# -----------------------------
def plot_box(ax, df, keep_mark, gene_ids, ycol, color):
    d = df[(df["keep_mark"]==keep_mark) & (df["gene_id"].isin(gene_ids))]
    sns.boxplot(
        data=d, x="perturbation_strength", y=ycol, order=order,
        gap=0.1, width=0.6, fliersize=0, linewidth=0.3,
        saturation=1, color=color, ax=ax
    )
    ax.set_xticks(order, xticklabels)
    ax.set_ylim(0, 1)
    ax.set_xlabel("")
    ax.tick_params(axis='x', width=0.5, labelsize=6)
    ax.tick_params(axis='y', width=0.5, labelsize=5)

def plot_count_bar(ax, df, keep_mark, baseline_label, flip_to_label, bar_color, ylabel, yticks=None, ylim=None):
    # 在 baseline（10 => 1.0）下属于 baseline_label 的基因集合
    base = df[(df["keep_mark"]==keep_mark) &
              (df["mean_label_label"]==baseline_label) &
              (df["mean_label_pred"]==baseline_label) &
              (df["perturbation_strength"]==10)]
    gene_ids = base["gene_id"].unique()

    # 计算每个扰动强度下，预测标签变为 flip_to_label 的基因数量
    d = df[(df["keep_mark"]==keep_mark) &
           (df["gene_id"].isin(gene_ids))]

    counts = (d[d["mean_label_pred"]==flip_to_label]
              .groupby("perturbation_strength", as_index=False)["gene_id"]
              .nunique()
              .rename(columns={"gene_id":"count"}))

    # 保证所有强度都有行（缺失补 0）
    all_levels = pd.DataFrame({"perturbation_strength": pd.Categorical(order, ordered=True, categories=order)})
    counts = all_levels.merge(counts, on="perturbation_strength", how="left").fillna({"count":0})

    sns.barplot(data=counts, x="perturbation_strength", y="count",
                order=order, color=bar_color, ax=ax, width=0.8)
    ax.set_xticks(order, xticklabels)
    ax.set_xlabel("")
    ax.set_ylabel(ylabel)
    if yticks is not None:
        ax.set_yticks(yticks)
    if ylim is not None:
        ax.set_ylim(ylim)

    return gene_ids  # 方便上游复用

# -----------------------------
# Figure layout
# -----------------------------
fig, ax = plt.subplots(3, 2, figsize=(7/2.54, 8/2.54), constrained_layout=True)

# ============ 左列：激活性 Mark（H3K27ac），低表达基因 ============
keep_mark = "H3K27ac"
baseline_label = 1   # “低表达”组
# 找出需要的 gene_id 列表（在 mean 表里）
# gene_ids_left = (df_mean[(df_mean["mean_label_label"]==baseline_label)]
#                  ["gene_id"].unique())
gene_ids_left = (df_mean[(df_mean["keep_mark"]==keep_mark) &
              (df_mean["mean_label_label"]==baseline_label) &
              (df_mean["mean_label_pred"]==baseline_label) &
              (df_mean["perturbation_strength"]==10)]["gene_id"].unique())

# BF
plot_box(ax[0,0], df_bfbs, keep_mark, gene_ids_left, ycol="bf_label_score", color=palette_box[0])
ax[0,0].set_ylabel("BF score")

# BS
plot_box(ax[1,0], df_bfbs, keep_mark, gene_ids_left, ycol="bs_label_score", color=palette_box[1])
ax[1,0].set_ylabel("BS score")

# 低表达 → 被扰动后“掉到 0 组”（即 flip_to_label=0）的基因数
plot_count_bar(
    ax=ax[2,0], df=df_mean, keep_mark=keep_mark,
    baseline_label=1, flip_to_label=0,
    bar_color=bar_low_color,
    ylabel="Number of lowly\nexpressed genes",
    yticks=[0,2000,4000,6000,8000], ylim=(0,8000)
)

# ============ 右列：抑制性 Mark（H3K27me3），高表达基因 ============
keep_mark = "H3K27me3"
baseline_label = 0   # “高表达”组在你的标注里用 0/1？这里沿用你原始逻辑：baseline_label=0
# gene_ids_right = (df_mean[(df_mean["mean_label_label"]==baseline_label)]
#                   ["gene_id"].unique())

gene_ids_right = (df_mean[(df_mean["keep_mark"]==keep_mark) &
              (df_mean["mean_label_label"]==baseline_label) &
              (df_mean["mean_label_pred"]==baseline_label) &
              (df_mean["perturbation_strength"]==10)]["gene_id"].unique())

# BF
plot_box(ax[0,1], df_bfbs, keep_mark, gene_ids_right, ycol="bf_label_score", color=palette_box[0])
ax[0,1].set_ylabel("BF score")

# BS
plot_box(ax[1,1], df_bfbs, keep_mark, gene_ids_right, ycol="bs_label_score", color=palette_box[1])
ax[1,1].set_ylabel("BS score")

# 高表达 → 被扰动后“翻到 1 组”（flip_to_label=1）的基因数
plot_count_bar(
    ax=ax[2,1], df=df_mean, keep_mark=keep_mark,
    baseline_label=0, flip_to_label=1,
    bar_color=bar_high_color,
    ylabel="Number of highly\nexpressed genes",
    yticks=[0,2000,4000,6000], ylim=(0,6000)
)

# 统一的 X 轴大标题
fig.supxlabel("Perturbation strength", fontsize=6)

# 保存与显示
# plt.savefig(save_path, format="eps", bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
from pathlib import Path
import re
from matplotlib.gridspec import GridSpec 

# ========= 路径 =========
EMISS_CSV   = "chromhmm_E003/E003_chromHMM50_emissions_7marks_prob.csv"  # 50×7，行=state，列=7 marks
PERTURB_CSV = "extra/datasets/benchmark/Promoterformer/results/perturbation_masks_mode-all.csv"
# PERTURB_CSV = "extra/datasets/benchmark/Promoterformer/results/perturbation_masks_mode-all-masks_strength-0.4.csv"
OUT_DIR     = Path("figures_chromhmm_panels")
OUT_DIR.mkdir(parents=True, exist_ok=True)

marks_order = [ 'H3K9ac', 'H3K27ac','H3K4me3','H3K4me1','H3K36me3','H3K27me3','H3K9me3']
# ========= 读取 =========
emiss = pd.read_csv(EMISS_CSV, index_col=0)[marks_order]      # 可能是 S1..S50 或其它格式
perturb = pd.read_csv(PERTURB_CSV)
perturb = perturb[perturb["eid"]=="E003"].copy() # 只看 E003

# ========= 统一 state 索引为 S1..S50 =========
def to_state_label(x):
    if isinstance(x, (int, np.integer)):
        return f"S{int(x)}"
    s = str(x)
    m = re.search(r'(\d+)', s)
    if not m:
        raise ValueError(f"Cannot parse state index: {x!r}")
    return f"S{int(m.group(1))}"

emiss.index = emiss.index.map(to_state_label)
desired_states = [f"S{i}" for i in range(1, 50+1)]
# 如果 emiss 缺少部分状态，不要 reindex（避免全空）——先保持原有行序，但若包含全部 50 个则按 S1..S50 排
if set(desired_states).issubset(set(emiss.index)):
    emiss = emiss.reindex(desired_states)

# ========= 组合规范化（顺序无关） =========
def combo_to_key(s):
    if s is None or (isinstance(s, float) and np.isnan(s)):
        return tuple()
    if isinstance(s, str):
        s = s.strip()
        if s=="" or s.lower()=="none":
            return tuple()
        for sep in [",", " "]:
            s = s.replace(sep, "|")
        parts = [p.strip() for p in s.split("|") if p.strip()]
        return tuple(sorted(parts))
    return tuple()

perturb["combo_key"] = perturb["masked_marks"].apply(combo_to_key)

# ========= emission → 组合键（阈值法） =========
THRESH_PRESENT = 0.1
def row_to_key(row, thresh=THRESH_PRESENT):
    present = [m for m,p in row.items() if p >= thresh]
    return tuple(sorted(present)) if present else tuple()

state2key = emiss.apply(row_to_key, axis=1)  # index=S*, value=tuple()

# ========= baseline 与 Δscore（不分正负样本）=========
baseline = perturb[perturb["combo_key"]==tuple()].set_index("gene_id")
for tgt in ["bf_label","bs_label"]:
    perturb[f"{tgt}_dscore"] = perturb[f"{tgt}_score"] - perturb["gene_id"].map(baseline[f"{tgt}_score"])

# ========= 按 state 聚合 Δscore 的均值 =========
grp = {k:g for k,g in perturb.groupby("combo_key")}
rows = []
for state, key in state2key.items():
    sub = grp.get(key, pd.DataFrame())
    rows.append({
        "state": state,
        "masked_marks": "|".join(key) if key else "None",
        "bf_mean": sub["bf_label_dscore"].mean() if not sub.empty else np.nan,
        "bs_mean": sub["bs_label_dscore"].mean() if not sub.empty else np.nan,
        "n_rows": len(sub)
    })
agg = pd.DataFrame(rows).set_index("state")

# 与 emission 对齐（若 emission 有完整 S1..S50）
if set(agg.index).issuperset(set(emiss.index)):
    agg = agg.reindex(emiss.index)

# ========= 新增：构造 ablation 组合矩阵（state × marks 的 0/1） =========
# marks_order = marks_order  # 与 emission 热图一致的列顺序
ablation_bin = pd.DataFrame(0, index=emiss.index, columns=marks_order, dtype=int)
for st, combo in state2key.items():
    if st in ablation_bin.index:
        for m in combo:
            if m in ablation_bin.columns:
                ablation_bin.loc[st, m] = 1

# ========= 画图函数：最左 ablation + 中 emission + 右柱状图（单 target） =========
def panel_ablate_emiss_bar(emiss_df, agg_df, ablate_df, target_col, title, out_prefix,sorted_delta_score=True):
    # 统一索引顺序（只用三者都存在的 state，按 emiss_df 的顺序）
    states = [s for s in emiss_df.index if (s in agg_df.index and s in ablate_df.index)]
    emiss_df  = emiss_df.loc[states]
    ablate_df = ablate_df.loc[states]
    sub = agg_df.loc[states, [target_col]].copy()

    n = len(states)
    y_centers = np.arange(n) + 0.5

    # ========= 排序：按 Δscore 绝对值从大到小（保持三者一致）=========
    sub["abs_val"] = sub[target_col].abs()
    if sorted_delta_score:
        sub = sub.sort_values("abs_val", ascending=False)
    states_sorted = sub.index.tolist()
    # emiss_df  = emiss_df.loc[states_sorted]
    # ablate_df = ablate_df.loc[states_sorted]

    n = len(states_sorted)
    y_centers = np.arange(n) + 0.5

    # Δscore 颜色映射（对称）
    v = sub[target_col].to_numpy()
    vmax = np.nanmax(np.abs(v)) if np.isfinite(v).any() else 0.5
    vmax = max(vmax, 1e-3)
    norm = mpl.colors.TwoSlopeNorm(vmin=-vmax, vcenter=0.0, vmax=vmax)
    cmap = mpl.cm.get_cmap("bwr")

    # 画布：左(黑点) | 中(emission) | 右(柱状) | gap | colorbar
    fig = plt.figure(figsize=(10.0/2.54, 8.0/2.54))
    gs = GridSpec(
        nrows=1, ncols=5,
        width_ratios=[4.2/2.54, 6.0/2.54, 4.2/2.54, 0.4/2.54, 0.25/2.54],
        wspace=0.55, figure=fig
    )

    ax_abl = fig.add_subplot(gs[0, 0])   # ablation 组合(黑点矩阵)
    ax_left = fig.add_subplot(gs[0, 1])  # emission 热图
    ax_bar  = fig.add_subplot(gs[0, 2])  # Δscore 柱状
    ax_gap  = fig.add_subplot(gs[0, 3])  # 空白
    cax1    = fig.add_subplot(gs[0, 4])  # emission colorbar
    ax_gap.axis("off")

    # ---- 最左：ablation 黑点矩阵 ----
    ax_abl.set_xlim(-0.5, len(marks_order)-0.5)
    ax_abl.set_ylim(0, n)
    ax_abl.invert_yaxis()
    ax_abl.set_xticks(np.arange(len(marks_order)))
    ax_abl.set_xticklabels(marks_order, rotation=90)
    ax_abl.set_yticks(y_centers)
    ax_abl.set_ylabel("States (1–50)")
    # 把 S1..S50 的前缀去掉，只显示数字
    ylab_nums = [s[1:] if isinstance(s, str) and s.startswith("S") else s for s in emiss_df.index]
    ax_abl.set_yticklabels(ylab_nums, fontsize=4)
    ax_abl.set_title("Masked marks\n(per state)")

    # 画黑点：1 的位置画一个点
    yy, xx = np.where(ablate_df.to_numpy()==1)
    ax_abl.scatter(xx, yy+0.5, s=3, c="#595757", marker="o")

    # ---- 中：emission 热图（与左侧同一 y 行）----
    hm = sns.heatmap(
        emiss_df, ax=ax_left, cmap="Reds", vmin=0, vmax=1,
        cbar=True, cbar_ax=cax1, yticklabels=True
    )
    ax_left.set_yticks([])
    ax_left.set_yticklabels([], fontsize=5)
    ax_left.set_ylim(0, n)
    ax_left.invert_yaxis()
    ax_left.set_xticks(np.arange(len(marks_order)))
    ax_left.set_xticklabels(marks_order, rotation=90)
    ax_left.set_title("H1 ChromHMM 50-state\nemissions (7 marks)")
    ax_left.set_xlabel("")
    ax_left.set_ylabel("")
    for lab in ax_left.get_xticklabels():
        lab.set_ha("right")    
    # 防止右侧标签被裁切，略扩 xlim
    nmarks = emiss_df.shape[1]
    ax_left.set_xlim(-0.5, nmarks - 0.5 + 0.6)
    cax1.set_ylabel("Emission probability", rotation=90, fontsize=7)

    # ---- 右：Δscore 柱状图（与左/中同一 y 行）----
    ax_bar.barh(y_centers, v, height=1.0, color=cmap(norm(v)))
    ax_bar.axvline(0, color="k", lw=0.5, alpha=0.6)
    ax_bar.set_ylim(0, n)
    ax_bar.set_yticks(y_centers)
    sub_states = sub.index.tolist()
    sub_states = [s[1:] if isinstance(s, str) and s.startswith("S") else s for s in sub_states]
    ax_bar.set_yticklabels(sub_states, fontsize=4)
    ax_bar.invert_yaxis()
    ax_bar.set_xlabel("Mean Δscore\n(perturb − baseline)")
    ax_bar.set_ylabel("")
    ax_bar.set_title(title)
    ax_bar.set_xlim(-0.5, 0.5)

    plt.tight_layout()
    fig.savefig(OUT_DIR / f"{out_prefix}.pdf", bbox_inches="tight")
    fig.savefig(OUT_DIR / f"{out_prefix}.eps", format="eps", bbox_inches="tight")
    plt.show()
    print("[Saved]", OUT_DIR / f"{out_prefix}.pdf")

# ========= 分别输出 BF / BS 两套图（带 ablation 面板）=========
panel_ablate_emiss_bar(
    emiss_df=emiss,
    agg_df=agg,
    ablate_df=ablation_bin,
    target_col="bf_mean",
    title="H1: Mean Δscore\nper state (BF, all genes)",
    out_prefix="H1_panel_ablate_emiss_bar_BF_no_delta_score",
    sorted_delta_score=False
)

panel_ablate_emiss_bar(
    emiss_df=emiss,
    agg_df=agg,
    ablate_df=ablation_bin,
    target_col="bs_mean",
    title="H1: Mean Δscore\nper state (BS, all genes)",
    out_prefix="H1_panel_ablate_emiss_bar_BS_no_delta_score",
    sorted_delta_score=False
)

# 同时保存汇总表
agg.to_csv(OUT_DIR / "E003_states_dscore_overall.csv")
print("[Saved table]", OUT_DIR / "E003_states_dscore_overall.csv")
```

```python
def panel_ablate_emiss_bar_dual_right(emiss_df, agg_df, ablate_df, out_prefix, sorted_delta_score=True):
    # 保证索引对齐
    states = [s for s in emiss_df.index if (s in agg_df.index and s in ablate_df.index)]
    emiss_df  = emiss_df.loc[states]
    ablate_df = ablate_df.loc[states]
    sub = agg_df.loc[states, ["bf_mean", "bs_mean"]].copy()

    # 排序
    sub["abs_val"] = sub[["bf_mean","bs_mean"]].abs().max(axis=1)
    if sorted_delta_score:
        sub = sub.sort_values("abs_val", ascending=False)
    states_sorted = sub.index.tolist()

    n = len(states_sorted)
    y_centers = np.arange(n) + 0.5

    # 配色
    vmax = np.nanmax(sub[["bf_mean","bs_mean"]].abs().to_numpy())
    vmax = max(vmax, 1e-3)
    norm = mpl.colors.TwoSlopeNorm(vmin=-vmax, vcenter=0.0, vmax=vmax)
    cmap = mpl.cm.get_cmap("bwr")

    # ===== 画布：左(ablation) | 中(BF,BS bar) | 右(emission+colorbar) =====
    fig = plt.figure(figsize=(11/2.54, 8/2.54))
    gs = GridSpec(
        nrows=1, ncols=6,
        width_ratios=[3.2/2.54, 3.2/2.54, 3.2/2.54, 3.2/2.54, 0.4/2.54, 0.25/2.54],
        wspace=0.5, figure=fig
    )

    ax_abl  = fig.add_subplot(gs[0,0])   # ablation
    ax_bar1 = fig.add_subplot(gs[0,1])   # BF bar
    ax_bar2 = fig.add_subplot(gs[0,2])   # BS bar
    ax_left = fig.add_subplot(gs[0,3])   # emission
    ax_gap  = fig.add_subplot(gs[0,4])   # 空白
    cax1    = fig.add_subplot(gs[0,5])   # colorbar
    ax_gap.axis("off")

    # ===== 最左：ablation =====
    ablate_df = ablate_df.loc[states_sorted]
    ax_abl.set_xlim(-0.5, len(marks_order)-0.5)
    ax_abl.set_ylim(0, n)
    ax_abl.invert_yaxis()
    ax_abl.set_xticks(np.arange(len(marks_order)))
    ax_abl.set_xticklabels(marks_order, rotation=90)
    ax_abl.set_yticks(y_centers)
    ax_abl.set_ylabel("States (1–50)")
    ylab_nums = [s[1:] if s.startswith("S") else s for s in states_sorted]
    ax_abl.set_yticklabels(ylab_nums, fontsize=4)
    ax_abl.set_title("Masked marks\n(per state)")
    yy, xx = np.where(ablate_df.to_numpy()==1)
    ax_abl.scatter(xx, yy+0.5, s=3, c="#595757", marker="o")

    # ===== bar 图：BF =====
    v1 = sub["bf_mean"].to_numpy()
    ax_bar1.barh(y_centers, v1, height=1.0, color=cmap(norm(v1)))
    ax_bar1.axvline(0, color="k", lw=0.5, alpha=0.6)
    ax_bar1.set_ylim(0, n)
    ax_bar1.set_yticks(y_centers)
    ax_bar1.set_yticklabels([], fontsize=4)
    ax_bar1.invert_yaxis()
    ax_bar1.set_title("Mean Δscore\n(BF)")
    ax_bar1.set_xlabel("Δscore")
    ax_bar1.set_xlim(-0.5, 0.5)

    # ===== bar 图：BS =====
    v2 = sub["bs_mean"].to_numpy()
    ax_bar2.barh(y_centers, v2, height=1.0, color=cmap(norm(v2)))
    ax_bar2.axvline(0, color="k", lw=0.5, alpha=0.6)
    ax_bar2.set_ylim(0, n)
    ax_bar2.set_yticks(y_centers)
    ax_bar2.set_yticklabels([], fontsize=4)
    ax_bar2.invert_yaxis()
    ax_bar2.set_title("Mean Δscore\n(BS)")
    ax_bar2.set_xlabel("Δscore")
    ax_bar2.set_xlim(-0.5, 0.5)

    # ===== emission =====
    emiss_df = emiss_df.loc[states_sorted]
    sns.heatmap(
        emiss_df, ax=ax_left, cmap="Reds", vmin=0, vmax=1,
        cbar=True, cbar_ax=cax1, yticklabels=False
    )
    ax_left.set_ylim(0, n)
    ax_left.invert_yaxis()
    ax_left.set_xticks(np.arange(len(marks_order)))
    ax_left.set_xticklabels(marks_order, rotation=90)
    ax_left.set_yticks(y_centers,labels=ylab_nums, fontsize=4)
    ax_left.set_ylabel("States")
    ax_left.set_title("H1 ChromHMM 50-state\nemissions (7 marks)",fontsize=5)
    cax1.set_ylabel("Emission probability", rotation=90)

    plt.tight_layout()
    fig.savefig(OUT_DIR / f"{out_prefix}_BF_BS_right.pdf", bbox_inches="tight")
    fig.savefig(OUT_DIR / f"{out_prefix}_BF_BS_right.eps", format="eps", bbox_inches="tight")
    plt.show()
    print("[Saved]", OUT_DIR / f"{out_prefix}_BF_BS_right.pdf")


# ===== 调用 =====
panel_ablate_emiss_bar_dual_right(
    emiss_df=emiss,
    agg_df=agg,
    ablate_df=ablation_bin,
    out_prefix="H1_panel_ablate_emiss_bar_dual_right",
    sorted_delta_score=False
)
```

```python

def bin_40kbp_matrix(matrix: np.ndarray, bin_size_bp: int = 500) -> np.ndarray:
    """
    将 40kbp × tracks 的矩阵，按 bin_size_bp 分箱平均。
    
    参数：
        matrix: np.ndarray, shape=(40000, C)，40kbp 范围的预测结果
        bin_size_bp: int, 每个 bin 的碱基长度，默认 500bp
    
    返回：
        np.ndarray, shape=(N_bins, C)，N_bins = 40000/bin_size_bp
    """
    length, C = matrix.shape
    assert length == 40000, f"输入长度应为40kbp，这里是 {length}"
    
    n_bins = length // bin_size_bp  # 80
    binned = matrix.reshape(n_bins, bin_size_bp, C).mean(axis=1)
    return binned


# 7 个 histone track 的名字（保持保存时的顺序）

def plot_tracks(tracks, ylim=None, is_binned=True,color_map = None,legend_order=None, figsize = (8/2.54, 3.5/2.54),legend = False,saved_path = None):
    """
    绘制多个组蛋白修饰的轨迹。

    参数：
        tracks (np.ndarray): 形状为 (N_bins, N_tracks) 的矩阵
        track_names (list): 轨迹名称列表
    """
    track_names = [
    'H3K4me1',
    'H3K4me3',
    'H3K9me3',
    'H3K27me3',
    'H3K36me3',
    'H3K27ac',
    'H3K9ac'
    ]
    if is_binned:
        tracks = bin_40kbp_matrix(tracks)
    # 横轴坐标：-20kb 到 +20kb
    x = (np.arange(tracks.shape[0]) - (tracks.shape[0] // 2 + 1)) * 500


    plt.figure(figsize=figsize)
    for i, name in enumerate(track_names):
        color = color_map[name] if color_map else None
        if color:
            plt.plot(x, tracks[:, i], label=name, lw=0.3, color=color)
        else:
            # 如果没有提供颜色，则使用默认颜色
            plt.plot(x, tracks[:, i], label=name, lw=0.3)

    plt.axvline(0, color='grey', linestyle='--', lw=0.5, label="TSS")

    if ylim:
       plt.ylim(top=ylim)
    plt.xlabel("Position relative to TSS (kbp)")
    plt.ylabel("Signal intensity\n(Enformer prediction)")
    handles, labels = plt.gca().get_legend_handles_labels()
    if legend_order:
        # 按 legend_order 重新排序
        order_dict = {name: i for i, name in enumerate(legend_order)}
        sorted_items = sorted(zip(handles, labels), key=lambda x: order_dict.get(x[1], 1e9))
        handles, labels = zip(*sorted_items)

    if legend:
        plt.legend(
            handles,
            labels,
            bbox_to_anchor=(1.1, 1.05),
            loc='upper left',
            fontsize=5
        )
    else:
        plt.legend().remove()
    plt.tight_layout()
    plt.xticks([-20000, 0, 20000], ['-20', '0', '20'])

    if saved_path:
        plt.savefig(saved_path, format='eps', bbox_inches='tight')
    plt.show()



```

```python
legend_order = ['H3K9ac','H3K27ac','H3K4me3','H3K4me1','H3K36me3','H3K27me3','H3K9me3']
tracks = np.load('extra/datasets/processed/v1/ENSG00000143340_random_center_replaced_100k/regions/ENSGRND00000100.npy') # (80, 7)
# tracks = np.load('extra/datasets/processed/v1/E003/regions/ENSG00000155660.npy')[:7].T # (80, 7)
# saved_path = 'extra/datasets/benchmark/Promoterformer/figures/v6/Figure5F.eps'

plot_tracks(tracks,ylim=60,color_map=COLOR_MAP,legend_order=legend_order,legend=True)
```

# Enformer 预测与Chip-seq 组蛋白修饰相关性

```python
import os
import pandas as pd 
from scipy.stats import pearsonr
from tqdm import tqdm
meta_data = pd.read_csv('extra/datasets/processed/v1/meta_datasets/meta_data_E003.csv')
```

```python
corr = 0
corr_samples = []
for gene_id in tqdm(meta_data['gene_id'].unique()):
    path1 = os.path.join('extra/datasets/processed/v1/E003/regions', f'{gene_id}.npy')
    path2 = os.path.join('extra/datasets/processed/v1/E003_enformer', f'{gene_id}.npy')
    
    if not os.path.exists(path1) or not os.path.exists(path2):
        continue
    
    data1 = np.load(path1)[:7].T
    data1 = bin_40kbp_matrix(data1, bin_size_bp=500)
    # data1 = data1[39:41,:]
    data2 = np.load(path2)
    data2 = bin_40kbp_matrix(data2, bin_size_bp=500)
    # data2 = data2[39:41,:]
    corr_sample = 0
    for idx in range(7):
         corr_sample += pearsonr(np.log1p(data1[:,idx]), np.log1p(data2[:,idx]))[0]    

    corr_sample /= 7
    if np.isnan(corr_sample):
        corr_sample = 0.0
    corr += corr_sample 
    corr_samples.append(corr_sample)
    # print(f"Gene: {gene_id}, Pearson correlation: {corr_sample:.4f}")
print(f"Average Pearson correlation: {corr / len(meta_data['gene_id'].unique()):.4f}")
    
```
