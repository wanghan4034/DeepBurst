```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

source_data_list = []

eid = 'E116'
cell_type = 'GM12878'
label = 1
keep_mark = 'H3K27ac'

# =========================================================
# X AXIS
# =========================================================
order = list(range(10, -1, -1))
xticklabels = [f"{1-i * 0.1:.1f}" if i % 2 == 0 else '' for i in order]

# =========================================================
# SELECT HIGH-EXPRESSION GENES
# =========================================================
data = pd.read_csv(f"extra/results/{eid}_perturbation_predictions_mean.csv")
data['perturbation_strength'] = data['perturbation_strength'].apply(lambda x: int(x * 10))
data = data[data['mean_label_label'] == label]
data = data[data['perturbation_strength'] == 10]

gene_ids = list(data['gene_id'].unique())
n_high_expression = len(gene_ids)
print(f'Fig. 5a: n={n_high_expression:,} high-expression genes in GM12878')
# =========================================================
# FIGURE
# =========================================================
fig, ax = plt.subplots(3, 2, figsize=(7/2.54, 8/2.54), constrained_layout=True)

palette = ['#E94C72', '#036EB8']

# =========================================================
# LEFT TOP / MIDDLE
# H3K27ac perturbation: BF / BS score
# =========================================================
data = pd.read_csv(f"extra/results/{eid}_perturbation_predictions_bs_bf.csv")
data['perturbation_strength_raw'] = data['perturbation_strength']
data['perturbation_strength'] = data['perturbation_strength'].apply(lambda x: int(x * 10))
data = data[(data['keep_mark'] == keep_mark) & (data['gene_id'].isin(gene_ids))].copy()
data['displayed_perturbation_strength'] = 1 - data['perturbation_strength'] / 10

# =========================================================
# SOURCE DATA: H3K27ac BF / BS
# =========================================================
source_panel = data.copy()
source_panel['source_type'] = 'score'
source_panel['figure_group'] = 'H3K27ac_high_expression'
source_panel['cell_type'] = cell_type
source_panel['expression_group'] = 'High-expression genes'
source_data_list.append(source_panel)

# =========================================================
# BF
# =========================================================
sns.boxplot(data=data, y='bf_label_score', x='perturbation_strength', order=order, gap=0.1, width=0.6, fliersize=0, linewidth=0.3, ax=ax[0, 0], color=palette[0], saturation=1)

ax[0, 0].set_xticks(order, labels=xticklabels)
ax[0, 0].set_ylim(0, 1)
ax[0, 0].set_xlabel("", fontsize=6)
ax[0, 0].set_ylabel("BF score", fontsize=6)

# =========================================================
# BS
# =========================================================
sns.boxplot(data=data, y='bs_label_score', x='perturbation_strength', order=order, gap=0.1, width=0.6, fliersize=0, linewidth=0.3, ax=ax[1, 0], color=palette[1], saturation=1)

ax[1, 0].set_xticks(order, labels=xticklabels)
ax[1, 0].set_ylim(0, 1)
ax[1, 0].set_xlabel("", fontsize=6)
ax[1, 0].set_ylabel("BS score", fontsize=6)

# =========================================================
# LEFT BOTTOM
# H3K27ac: HIGH -> LOW EXPRESSION
# =========================================================
label = 1
keep_mark = 'H3K27ac'
color = '#ECD698'

data = pd.read_csv(f"extra/results/{eid}_perturbation_predictions_mean.csv")
data['perturbation_strength_raw'] = data['perturbation_strength']
data['perturbation_strength'] = data['perturbation_strength'].apply(lambda x: int(x * 10))
data = data[(data['mean_label_label'] == label) & (data['keep_mark'] == keep_mark)].copy()

baseline_data = data[(data['mean_label_label'] == label) & (data['perturbation_strength'] == 10) & (data['mean_label_pred'] == label) & (data['keep_mark'] == keep_mark)]
gene_ids = baseline_data['gene_id']

print(len(gene_ids))

# =========================================================
# SOURCE DATA: H3K27ac expression state
# =========================================================
source_panel = data[data['gene_id'].isin(gene_ids)].copy()
source_panel['source_type'] = 'expression_state'
source_panel['figure_group'] = 'H3K27ac_high_to_low'
source_panel['cell_type'] = cell_type
source_panel['expression_group'] = 'Baseline high-expression genes'
source_panel['displayed_perturbation_strength'] = 1 - source_panel['perturbation_strength'] / 10
source_data_list.append(source_panel)

positive_genes = data[(data['gene_id'].isin(gene_ids)) & (data['mean_label_pred'] == 0)][['perturbation_strength', 'gene_id']].groupby('perturbation_strength', as_index=False).count()
positive_genes = positive_genes.sort_values(by='perturbation_strength', ascending=False)

print(positive_genes)

order = list(range(10, -1, -1))
negative_genes_nums = [0] + list(positive_genes['gene_id'])

sns.barplot(negative_genes_nums, color=color, ax=ax[2, 0])

ax[2, 0].set_xticks(order, labels=xticklabels)
ax[2, 0].set_ylabel('The number of lowly\nexpressed genes (×10³)')
ax[2, 0].set_xlabel("", fontsize=6)
ax[2, 0].set_yticklabels([0, 2, 4, 6, 8])
ax[2, 0].set_ylim([0, 8000])

# =========================================================
# RIGHT
# H3K27me3 perturbation
# =========================================================
keep_mark = 'H3K27me3'
label = 0

order = list(range(10, -1, -1))
xticklabels = [f"{1-i * 0.1:.1f}" if i % 2 == 0 else '' for i in order]

# =========================================================
# SELECT LOW-EXPRESSION GENES
# =========================================================
data = pd.read_csv(f"extra/results/{eid}_perturbation_predictions_mean.csv")
data['perturbation_strength'] = data['perturbation_strength'].apply(lambda x: int(x * 10))
data = data[data['mean_label_label'] == label]
data = data[data['perturbation_strength'] > 0]

gene_ids = list(data['gene_id'].unique())
n_low_expression = len(gene_ids)
print(f'Fig. 5b: n={n_low_expression:,} low-expression genes in GM12878')

palette = ['#E94C72', '#036EB8']

# =========================================================
# RIGHT TOP / MIDDLE
# H3K27me3 perturbation: BF / BS score
# =========================================================
data = pd.read_csv(f"extra/results/{eid}_perturbation_predictions_bs_bf.csv")
data['perturbation_strength_raw'] = data['perturbation_strength']
data['perturbation_strength'] = data['perturbation_strength'].apply(lambda x: int(x * 10))
data = data[(data['keep_mark'] == keep_mark) & (data['gene_id'].isin(gene_ids))].copy()
data['displayed_perturbation_strength'] = 1 - data['perturbation_strength'] / 10

# =========================================================
# SOURCE DATA: H3K27me3 BF / BS
# =========================================================
source_panel = data.copy()
source_panel['source_type'] = 'score'
source_panel['figure_group'] = 'H3K27me3_low_expression'
source_panel['cell_type'] = cell_type
source_panel['expression_group'] = 'Low-expression genes'
source_data_list.append(source_panel)

# =========================================================
# BF
# =========================================================
sns.boxplot(data=data, y='bf_label_score', x='perturbation_strength', order=order, gap=0.1, width=0.6, fliersize=0, linewidth=0.3, ax=ax[0, 1], color=palette[0], saturation=1)

ax[0, 1].set_xticks(order, labels=xticklabels)
ax[0, 1].set_ylim(0, 1)
ax[0, 1].set_xlabel("", fontsize=6)
ax[0, 1].set_ylabel("BF score", fontsize=6)

# =========================================================
# BS
# =========================================================
sns.boxplot(data=data, y='bs_label_score', x='perturbation_strength', order=order, gap=0.1, width=0.6, fliersize=0, linewidth=0.3, ax=ax[1, 1], color=palette[1], saturation=1)

ax[1, 1].set_xticks(order, labels=xticklabels)
ax[1, 1].set_ylim(0, 1)
ax[1, 1].set_xlabel("", fontsize=6)
ax[1, 1].set_ylabel("BS score", fontsize=6)

# =========================================================
# RIGHT BOTTOM
# H3K27me3: LOW -> HIGH EXPRESSION
# =========================================================
label = 0
keep_mark = 'H3K27me3'
color = '#E2AC91'

data = pd.read_csv(f"extra/results/{eid}_perturbation_predictions_mean.csv")
data['perturbation_strength_raw'] = data['perturbation_strength']
data['perturbation_strength'] = data['perturbation_strength'].apply(lambda x: int(x * 10))
data = data[(data['mean_label_label'] == label) & (data['keep_mark'] == keep_mark)].copy()

baseline_data = data[(data['mean_label_label'] == label) & (data['perturbation_strength'] == 10) & (data['mean_label_pred'] == label) & (data['keep_mark'] == keep_mark)]
gene_ids = baseline_data['gene_id']

print(len(gene_ids))

# =========================================================
# SOURCE DATA: H3K27me3 expression state
# =========================================================
source_panel = data[data['gene_id'].isin(gene_ids)].copy()
source_panel['source_type'] = 'expression_state'
source_panel['figure_group'] = 'H3K27me3_low_to_high'
source_panel['cell_type'] = cell_type
source_panel['expression_group'] = 'Baseline low-expression genes'
source_panel['displayed_perturbation_strength'] = 1 - source_panel['perturbation_strength'] / 10
source_data_list.append(source_panel)

positive_genes = data[(data['gene_id'].isin(gene_ids)) & (data['mean_label_pred'] == 1)][['perturbation_strength', 'gene_id']].groupby('perturbation_strength', as_index=False).count()
positive_genes = positive_genes.sort_values(by='perturbation_strength', ascending=False)

print(positive_genes)

order = list(range(10, -1, -1))
positive_genes_nums = [0] + list(positive_genes['gene_id']) + [0]

sns.barplot(positive_genes_nums, color=color, ax=ax[2, 1])

ax[2, 1].set_xticks(order, labels=xticklabels)
ax[2, 1].set_yticklabels([0, 2, 4, 6])
ax[2, 1].set_ylim([0, 6000])
ax[2, 1].set_ylabel('The number of highly\nexpressed genes (×10³)')
ax[2, 1].set_xlabel("", fontsize=6)

# =========================================================
# SOURCE DATA
# =========================================================
source_data = pd.concat(source_data_list, ignore_index=True, sort=False)

source_columns = [
    'source_type',
    'figure_group',
    'cell_type',
    'gene_id',
    'keep_mark',
    'expression_group',
    'perturbation_strength_raw',
    'perturbation_strength',
    'displayed_perturbation_strength',
    'fold',
    'bf_label_label',
    'bf_label_score',
    'bs_label_label',
    'bs_label_score',
    'mean_label_label',
    'mean_label_pred'
]

source_data = source_data[[col for col in source_columns if col in source_data.columns]]

source_dir = 'scripts/figures/v8/source_data'
os.makedirs(source_dir, exist_ok=True)
source_data.to_csv(f'{source_dir}/fig5a.csv', index=False)

print(f'Source data saved to {source_dir}/fig5a.csv')

print('\nSource data counts:')
print(source_data.groupby(['source_type', 'figure_group']).size())

# =========================================================
# FINAL LAYOUT
# =========================================================
fig.supxlabel("Perturbation strength", fontsize=6)

plt.savefig('extra/figures/v8/Figure5A.eps', format='eps', bbox_inches='tight')
plt.show()
```

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
from pathlib import Path
from matplotlib.gridspec import GridSpec


# =========================================================
# PATHS
# =========================================================
EMISS_CSV = "extra/results/chromhmm_E003/E003_chromHMM50_emissions_7marks_prob.csv"
PERTURB_CSV = "extra/results/perturbation_masks_mode-all-masks_strength-0.csv"

SOURCE_DIR = Path("scripts/figures/v8/source_data")
SOURCE_DIR.mkdir(parents=True, exist_ok=True)

marks_order = ['H3K9ac', 'H3K27ac', 'H3K4me3', 'H3K4me1', 'H3K36me3', 'H3K27me3', 'H3K9me3']


# =========================================================
# LOAD DATA
# =========================================================
emiss = pd.read_csv(EMISS_CSV, index_col=0)[marks_order]

perturb = pd.read_csv(PERTURB_CSV)
perturb = perturb[perturb['eid'] == 'E003'].copy()


# =========================================================
# STANDARDIZE STATE LABELS: S1-S50
# =========================================================
def to_state_label(x):
    if isinstance(x, (int, np.integer)):
        return f"S{int(x)}"

    s = str(x)
    m = re.search(r'(\d+)', s)

    if not m:
        raise ValueError(f"Cannot parse state index: {x!r}")

    return f"S{int(m.group(1))}"


emiss.index = emiss.index.map(to_state_label)

desired_states = [f"S{i}" for i in range(1, 51)]

if set(desired_states).issubset(set(emiss.index)):
    emiss = emiss.reindex(desired_states)


# =========================================================
# NORMALIZE MASKED-MARK COMBINATIONS
# =========================================================
def combo_to_key(s):
    if s is None or (isinstance(s, float) and np.isnan(s)):
        return tuple()

    if isinstance(s, str):
        s = s.strip()

        if s == "" or s.lower() == "none":
            return tuple()

        for sep in [",", " "]:
            s = s.replace(sep, "|")

        parts = [p.strip() for p in s.split("|") if p.strip()]
        return tuple(sorted(parts))

    return tuple()


perturb['combo_key'] = perturb['masked_marks'].apply(combo_to_key)


# =========================================================
# ChromHMM EMISSION -> MASKED-MARK COMBINATION
# =========================================================
THRESH_PRESENT = 0.1


def row_to_key(row, thresh=THRESH_PRESENT):
    present = [mark for mark, prob in row.items() if prob >= thresh]
    return tuple(sorted(present)) if present else tuple()


state2key = emiss.apply(row_to_key, axis=1)


# =========================================================
# BASELINE AND DELTA SCORES
# =========================================================
baseline = perturb[perturb['combo_key'] == tuple()].set_index('gene_id')

for target in ['bf_label', 'bs_label']:
    perturb[f'{target}_dscore'] = perturb[f'{target}_score'] - perturb['gene_id'].map(baseline[f'{target}_score'])


# =========================================================
# AGGREGATE BF / BS DELTA SCORE BY STATE
# =========================================================
groups = {key: group for key, group in perturb.groupby('combo_key')}

rows = []

for state, key in state2key.items():
    sub = groups.get(key, pd.DataFrame())

    rows.append({
        'state': state,
        'masked_marks': '|'.join(key) if key else 'None',
        'bf_mean': sub['bf_label_dscore'].mean() if not sub.empty else np.nan,
        'bs_mean': sub['bs_label_dscore'].mean() if not sub.empty else np.nan,
        'n_rows': len(sub)
    })

agg = pd.DataFrame(rows).set_index('state')

if set(agg.index).issuperset(set(emiss.index)):
    agg = agg.reindex(emiss.index)


# =========================================================
# BINARY MASK MATRIX
# =========================================================
ablation_bin = pd.DataFrame(0, index=emiss.index, columns=marks_order, dtype=int)

for state, combo in state2key.items():
    if state in ablation_bin.index:
        for mark in combo:
            if mark in ablation_bin.columns:
                ablation_bin.loc[state, mark] = 1


# =========================================================
# SOURCE DATA
# One row = one ChromHMM state
# =========================================================
source_data = agg.reset_index().copy()
source_data['cell_type'] = 'H1'

# ChromHMM emission probability
for mark in marks_order:
    source_data[f'emission_{mark}'] = source_data['state'].map(emiss[mark])

# Binary masked-mark indicator
for mark in marks_order:
    source_data[f'masked_{mark}'] = source_data['state'].map(ablation_bin[mark])

source_columns = ['state', 'cell_type', 'masked_marks', 'n_rows', 'bf_mean', 'bs_mean']

for mark in marks_order:
    source_columns.append(f'emission_{mark}')

for mark in marks_order:
    source_columns.append(f'masked_{mark}')

source_data = source_data[source_columns]

source_data.to_csv(SOURCE_DIR / 'fig5e.csv', index=False)

print(f"[Saved source data] {SOURCE_DIR / 'fig5e.csv'}")


# =========================================================
# FIGURE 5E
# =========================================================
def panel_ablate_emiss_bar_dual_right(emiss_df, agg_df, ablate_df, save_path, sorted_delta_score=True):

    # -----------------------------------------------------
    # ALIGN STATES
    # -----------------------------------------------------
    states = [state for state in emiss_df.index if state in agg_df.index and state in ablate_df.index]

    emiss_df = emiss_df.loc[states].copy()
    ablate_df = ablate_df.loc[states].copy()
    sub = agg_df.loc[states, ['bf_mean', 'bs_mean']].copy()

    # -----------------------------------------------------
    # OPTIONAL SORTING
    # -----------------------------------------------------
    sub['abs_val'] = sub[['bf_mean', 'bs_mean']].abs().max(axis=1)

    if sorted_delta_score:
        sub = sub.sort_values('abs_val', ascending=False)

    states_sorted = sub.index.tolist()

    emiss_df = emiss_df.loc[states_sorted]
    ablate_df = ablate_df.loc[states_sorted]

    n = len(states_sorted)
    y_centers = np.arange(n) + 0.5

    # -----------------------------------------------------
    # STATE LABELS
    # -----------------------------------------------------
    state_labels = [state[1:] if isinstance(state, str) and state.startswith('S') else state for state in states_sorted]

    # -----------------------------------------------------
    # FIGURE
    # masked marks | BF | BS | emission | gap | colorbar
    # -----------------------------------------------------
    fig = plt.figure(figsize=(11/2.54, 8/2.54))

    gs = GridSpec(nrows=1, ncols=6, width_ratios=[3.2/2.54, 3.2/2.54, 3.2/2.54, 3.2/2.54, 0.4/2.54, 0.25/2.54], wspace=0.5, figure=fig)

    ax_abl = fig.add_subplot(gs[0, 0])
    ax_bar1 = fig.add_subplot(gs[0, 1])
    ax_bar2 = fig.add_subplot(gs[0, 2])
    ax_emiss = fig.add_subplot(gs[0, 3])
    ax_gap = fig.add_subplot(gs[0, 4])
    cax = fig.add_subplot(gs[0, 5])

    ax_gap.axis('off')

    # =====================================================
    # MASKED HISTONE MARKS
    # =====================================================
    ax_abl.set_xlim(-0.5, len(marks_order) - 0.5)
    ax_abl.set_ylim(0, n)
    ax_abl.invert_yaxis()

    ax_abl.set_xticks(np.arange(len(marks_order)))
    ax_abl.set_xticklabels(marks_order, rotation=90, fontsize=5.5)

    ax_abl.set_yticks(y_centers)
    ax_abl.set_yticklabels(state_labels, fontsize=4)

    ax_abl.set_ylabel('ChromHMM States (1-50)')
    ax_abl.set_title('Masked histone marks\n(per perturbation)')

    yy, xx = np.where(ablate_df.to_numpy() == 1)
    ax_abl.scatter(xx, yy + 0.5, s=3, c='#595757', marker='o')

    # =====================================================
    # BF DELTA SCORE
    # =====================================================
    v1 = sub['bf_mean'].to_numpy()

    ax_bar1.barh(y_centers, v1, height=1.0, color=np.where(v1 >= 0, '#FF8484', '#5C8BFF'))
    ax_bar1.axvline(0, color='black', lw=0.5)

    ax_bar1.set_ylim(0, n)
    ax_bar1.set_yticks(y_centers)
    ax_bar1.set_yticklabels([])
    ax_bar1.invert_yaxis()

    ax_bar1.set_title('Mean Δscore\n(BF)')
    ax_bar1.set_xlabel('Δscore')
    ax_bar1.set_xlim(-0.5, 0.5)

    # =====================================================
    # BS DELTA SCORE
    # =====================================================
    v2 = sub['bs_mean'].to_numpy()

    ax_bar2.barh(y_centers, v2, height=1.0, color=np.where(v2 >= 0, '#FF8484', '#5C8BFF'))
    ax_bar2.axvline(0, color='black', lw=0.5)

    ax_bar2.set_ylim(0, n)
    ax_bar2.set_yticks(y_centers)
    ax_bar2.set_yticklabels([])
    ax_bar2.invert_yaxis()

    ax_bar2.set_title('Mean Δscore\n(BS)')
    ax_bar2.set_xlabel('Δscore')
    ax_bar2.set_xlim(-0.5, 0.5)

    # =====================================================
    # ChromHMM EMISSION HEATMAP
    # =====================================================
    sns.heatmap(emiss_df, ax=ax_emiss, cmap='Reds', vmin=0, vmax=1, cbar=True, cbar_ax=cax, yticklabels=False)

    ax_emiss.set_ylim(0, n)
    ax_emiss.invert_yaxis()

    ax_emiss.set_xticks(np.arange(len(marks_order)))
    ax_emiss.set_xticklabels(marks_order, rotation=90, fontsize=5.5)

    ax_emiss.set_yticks(y_centers, labels=state_labels, fontsize=4)

    ax_emiss.set_ylabel('States')
    ax_emiss.set_title('H1 ChromHMM 50-state\nemissions (7 marks)')

    cax.set_ylabel('Emission probability', rotation=90)

    # =====================================================
    # SAVE
    # =====================================================
    plt.tight_layout()

    fig.savefig(save_path, format='eps', bbox_inches='tight', dpi=300)

    plt.show()

    print(f'[Saved] {save_path}')


# =========================================================
# DRAW FIGURE 5E
# =========================================================
save_path = 'extra/figures/v7/Figure5E.eps'

panel_ablate_emiss_bar_dual_right(emiss_df=emiss, agg_df=agg, ablate_df=ablation_bin, save_path=save_path, sorted_delta_score=False)
```

```python

gene_id = "ENSG00000155660"  # for debugging.
npy_dir = f"extra/datasets/processed/v1/{gene_id}_random_center_replaced_100k/regions"
def bin_40kbp_matrix(matrix: np.ndarray, bin_size_bp: int = 500) -> np.ndarray:
    """
    将 40kbp × tracks 的矩阵，按 bin_size_bp 分箱平均。
    
    参数：
        matrix: np.ndarray, shape=(40000, C)，40kbp 范围的预测结果
        bin_size_bp: int, 每个 bin 的碱基长度，默认 500bp
    
    返回：
        np.ndarray, shape=(N_bins, C)，N_bins = 40000/bin_size_bp
    """
    length, C = matrix.shape
    assert length == 40000, f"输入长度应为40kbp，这里是 {length}"
    
    n_bins = length // bin_size_bp  # 80
    binned = matrix.reshape(n_bins, bin_size_bp, C).mean(axis=1)
    return binned


# 7 个 histone track 的名字（保持保存时的顺序）

def plot_tracks(tracks, ylim=None, is_binned=True,color_map = None,legend_order=None, figsize = (8/2.54, 3.5/2.54),legend = False,saved_path = None):
    """
    绘制多个组蛋白修饰的轨迹。

    参数：
        tracks (np.ndarray): 形状为 (N_bins, N_tracks) 的矩阵
        track_names (list): 轨迹名称列表
    """
    track_names = [
    'H3K4me1',
    'H3K4me3',
    'H3K9me3',
    'H3K27me3',
    'H3K36me3',
    'H3K27ac',
    'H3K9ac'
    ]
    if is_binned:
        tracks = bin_40kbp_matrix(tracks)
    # 横轴坐标：-20kb 到 +20kb
    x = (np.arange(tracks.shape[0]) - (tracks.shape[0] // 2 + 1)) * 500


    plt.figure(figsize=figsize)
    for i, name in enumerate(track_names):
        color = color_map[name] if color_map else None
        if color:
            plt.plot(x, tracks[:, i], label=name, lw=0.3, color=color)
        else:
            # 如果没有提供颜色，则使用默认颜色
            plt.plot(x, tracks[:, i], label=name, lw=0.3)

    plt.axvline(0, color='grey', linestyle='--', lw=0.5, label="TSS")

    if ylim:
       plt.ylim(top=ylim)
    plt.xlabel("Position relative to TSS (kbp)")
    plt.ylabel("Signal intensity\n(Enformer prediction)")
    handles, labels = plt.gca().get_legend_handles_labels()
    if legend_order:
        # 按 legend_order 重新排序
        order_dict = {name: i for i, name in enumerate(legend_order)}
        sorted_items = sorted(zip(handles, labels), key=lambda x: order_dict.get(x[1], 1e9))
        handles, labels = zip(*sorted_items)

    if legend:
        plt.legend(
            handles,
            labels,
            bbox_to_anchor=(1.1, 1.05),
            loc='upper left',
            fontsize=5.5,
            edgecolor='none',
        )
    else:
        plt.legend().remove()
    plt.tight_layout()
    plt.xticks([-20000, 0, 20000], ['-20', '0', '20'])

    if saved_path:
        plt.savefig(saved_path, format='eps', bbox_inches='tight', dpi=300)
    plt.show()



```

```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# =========================================================
# BASIC SETTINGS
# =========================================================
gene_id = "ENSG00000155660"
npy_dir = f"extra/datasets/processed/v1/{gene_id}_random_center_replaced_100k/regions"

track_names = [
    'H3K4me1',
    'H3K4me3',
    'H3K9me3',
    'H3K27me3',
    'H3K36me3',
    'H3K27ac',
    'H3K9ac'
]

legend_order = [
    'H3K9ac',
    'H3K27ac',
    'H3K4me3',
    'H3K4me1',
    'H3K36me3',
    'H3K27me3',
    'H3K9me3'
]

source_data_list = []


# =========================================================
# BINNING
# =========================================================
def bin_40kbp_matrix(matrix: np.ndarray, bin_size_bp: int = 500) -> np.ndarray:
    """
    Average a 40-kbp × tracks matrix into fixed-width bins.
    """
    length, C = matrix.shape
    assert length == 40000, f"输入长度应为40kbp，这里是 {length}"

    n_bins = length // bin_size_bp
    binned = matrix.reshape(n_bins, bin_size_bp, C).mean(axis=1)

    return binned


# =========================================================
# PREPARE SOURCE DATA
# =========================================================
def tracks_to_source_data(tracks, panel, sample_id, sample_type, is_binned=True):
    """
    Convert the exact data used for plotting into long-format source data.
    """

    if is_binned:
        tracks_plot = bin_40kbp_matrix(tracks)
    else:
        tracks_plot = tracks.copy()

    # Keep exactly the same x-coordinate definition as the plotting code
    x = (np.arange(tracks_plot.shape[0]) - (tracks_plot.shape[0] // 2 + 1)) * 500

    records = []

    for i, mark in enumerate(track_names):
        for j in range(tracks_plot.shape[0]):
            records.append({
                'panel': panel,
                'sample_id': sample_id,
                'sample_type': sample_type,
                'gene_id': gene_id,
                'histone_mark': mark,
                'bin_index': j,
                'position_bp': x[j],
                'position_kbp': x[j] / 1000,
                'signal_intensity': tracks_plot[j, i]
            })

    return pd.DataFrame(records)


# =========================================================
# PLOTTING FUNCTION
# =========================================================
def plot_tracks(tracks, ylim=None, is_binned=True, color_map=None, legend_order=None, figsize=(8/2.54, 3.5/2.54), legend=False, saved_path=None):

    if is_binned:
        tracks = bin_40kbp_matrix(tracks)

    # Horizontal position used in the actual figure
    x = (np.arange(tracks.shape[0]) - (tracks.shape[0] // 2 + 1)) * 500

    plt.figure(figsize=figsize)

    for i, name in enumerate(track_names):
        color = color_map[name] if color_map else None

        if color:
            plt.plot(x, tracks[:, i], label=name, lw=0.3, color=color)
        else:
            plt.plot(x, tracks[:, i], label=name, lw=0.3)

    plt.axvline(0, color='grey', linestyle='--', lw=0.5, label='TSS')

    if ylim:
        plt.ylim(top=ylim)

    plt.xlabel("Position relative to TSS (kbp)")
    plt.ylabel("Signal intensity\n(Enformer prediction)")

    handles, labels = plt.gca().get_legend_handles_labels()

    if legend_order:
        order_dict = {name: i for i, name in enumerate(legend_order)}
        sorted_items = sorted(zip(handles, labels), key=lambda x: order_dict.get(x[1], 1e9))
        handles, labels = zip(*sorted_items)

    if legend:
        plt.legend(handles, labels, bbox_to_anchor=(1.1, 1.05), loc='upper left', fontsize=5.5, edgecolor='none')
    else:
        legend_obj = plt.gca().get_legend()
        if legend_obj is not None:
            legend_obj.remove()

    plt.tight_layout()
    plt.xticks([-20000, 0, 20000], ['-20', '0', '20'])

    if saved_path:
        plt.savefig(saved_path, format='eps', bbox_inches='tight', dpi=300)

    plt.show()


# =========================================================
# FIGURE 5H
# ORIGINAL PDIA4 / H1 TRACKS
# =========================================================
tracks = np.load(f'extra/datasets/processed/v1/E003/regions/{gene_id}.npy')[:7].T

source_data_list.append(
    tracks_to_source_data(
        tracks=tracks,
        panel='Fig5h',
        sample_id=gene_id,
        sample_type='Original gene'
    )
)

saved_path = 'extra/figures/v8/Figure5H.eps'

plot_tracks(
    tracks,
    ylim=60,
    color_map=COLOR_MAP,
    legend_order=legend_order,
    legend=True,
    saved_path=saved_path
)


# =========================================================
# FIGURE 5I-1
# =========================================================
simulation_id = 'ENSGRND00003522'

if 'df' in globals():
    print(df[df['simulation_id'] == simulation_id])

tracks = np.load(f'{npy_dir}/{simulation_id}.npy')

source_data_list.append(
    tracks_to_source_data(
        tracks=tracks,
        panel='Fig5i-1',
        sample_id=simulation_id,
        sample_type='Random simulation'
    )
)

saved_path = 'extra/figures/v8/Figure5I_1.eps'
figsize = (4/2.54, 3.5/2.54)

plot_tracks(
    tracks,
    ylim=60,
    color_map=COLOR_MAP,
    legend_order=legend_order,
    figsize=figsize,
    saved_path=saved_path
)


# =========================================================
# FIGURE 5I-2
# =========================================================
simulation_id = 'ENSGRND00001485'

if 'df' in globals():
    print(df[df['simulation_id'] == simulation_id])

tracks = np.load(f'{npy_dir}/{simulation_id}.npy')

source_data_list.append(
    tracks_to_source_data(
        tracks=tracks,
        panel='Fig5i-2',
        sample_id=simulation_id,
        sample_type='Random simulation'
    )
)

saved_path = 'extra/figures/v8/Figure5I_2.eps'
figsize = (4/2.54, 3.5/2.54)

plot_tracks(
    tracks,
    ylim=60,
    color_map=COLOR_MAP,
    legend_order=legend_order,
    figsize=figsize,
    saved_path=saved_path
)


# =========================================================
# FIGURE 5I-3
# =========================================================
simulation_id = 'ENSGRND00002638'

if 'df' in globals():
    print(df[df['simulation_id'] == simulation_id])

tracks = np.load(f'{npy_dir}/{simulation_id}.npy')

source_data_list.append(
    tracks_to_source_data(
        tracks=tracks,
        panel='Fig5i-3',
        sample_id=simulation_id,
        sample_type='Random simulation'
    )
)

saved_path = 'extra/figures/v8/Figure5I_3.eps'
figsize = (4/2.54, 3.5/2.54)

plot_tracks(
    tracks,
    ylim=60,
    color_map=COLOR_MAP,
    legend_order=legend_order,
    figsize=figsize,
    saved_path=saved_path
)


# =========================================================
# FIGURE 5I-4
# =========================================================
simulation_id = 'ENSGRND00001752'

if 'df' in globals():
    print(df[df['simulation_id'] == simulation_id])

tracks = np.load(f'{npy_dir}/{simulation_id}.npy')

source_data_list.append(
    tracks_to_source_data(
        tracks=tracks,
        panel='Fig5i-4',
        sample_id=simulation_id,
        sample_type='Random simulation'
    )
)

saved_path = 'extra/figures/v8/Figure5I_4.eps'
figsize = (4/2.54, 3.5/2.54)

plot_tracks(
    tracks,
    ylim=60,
    color_map=COLOR_MAP,
    legend_order=legend_order,
    figsize=figsize,
    saved_path=saved_path
)


# =========================================================
# COMBINE FIGURE 5H + 5I SOURCE DATA
# =========================================================
source_data = pd.concat(source_data_list, ignore_index=True)

source_columns = [
    'panel',
    'sample_id',
    'sample_type',
    'gene_id',
    'histone_mark',
    'bin_index',
    'position_bp',
    'position_kbp',
    'signal_intensity'
]

source_data = source_data[source_columns]

source_dir = 'scripts/figures/v8/source_data'
os.makedirs(source_dir, exist_ok=True)

source_data.to_csv(f'{source_dir}/fig5hi.csv', index=False)

print(f'Source data saved to {source_dir}/fig5hi.csv')
print(source_data.groupby(['panel', 'sample_id', 'histone_mark']).size())
```

```python
corr = 0
corr_samples = []
for gene_id in tqdm(meta_data['gene_id'].unique()):
    path1 = os.path.join('extra/datasets/processed/v1/E003/regions', f'{gene_id}.npy')
    path2 = os.path.join('extra/datasets/processed/v2/E003_enformer/regions', f'{gene_id}.npy')
    
    if not os.path.exists(path1) or not os.path.exists(path2):
        continue
    
    data1 = np.load(path1)[:7].T
    data1 = bin_40kbp_matrix(data1, bin_size_bp=500)
    # data1 = data1[39:41,:]
    data2 = np.load(path2)
    data2 = bin_40kbp_matrix(data2, bin_size_bp=500)
    # data2 = data2[39:41,:]
    corr_sample = 0
    for idx in range(7):
         corr_sample += pearsonr(np.log1p(data1[:,idx]), np.log1p(data2[:,idx]))[0]    

    corr_sample /= 7
    if np.isnan(corr_sample):
        corr_sample = 0.0
    corr += corr_sample 
    corr_samples.append(corr_sample)
    # print(f"Gene: {gene_id}, Pearson correlation: {corr_sample:.4f}")
print(f"Average Pearson correlation: {corr / len(meta_data['gene_id'].unique()):.4f}")
    
```
