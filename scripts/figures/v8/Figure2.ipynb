```python
import matplotlib.patches as mpatches
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    1, 2,
    constrained_layout=True,
    figsize=(7/2.54, 3.5/2.54)
)

order = pd.Index(["E003","E118","E116"], name='marks')
hue_order = ['bf_label','bs_label']
palette = ['#E94C72','#036EB8']

x_pos = np.arange(len(order))
width = 0.28

# =========================================================
# GLOBAL LAYOUT CONTROL
# =========================================================
hue_spacing = 0.35
offsets = np.linspace(-hue_spacing/2, hue_spacing/2, len(hue_order))

# =========================================================
# LEFT PANEL (txburst)
# =========================================================
data = pd.read_csv('extra/results/bs_bf_para_indicators.csv')
data['auc'] = data['auc'] / 100

summary = data.groupby(['eid', 'target'])['auc'].agg(['mean', 'std']).reset_index()

# -------------------------
# BAR + ERRORBAR
# -------------------------
for i, tgt in enumerate(hue_order):
    sub = summary[summary['target'] == tgt]

    means = np.array([sub[sub['eid'] == eid]['mean'].values[0] for eid in order])
    stds  = np.array([sub[sub['eid'] == eid]['std'].values[0] for eid in order])

    xpos = x_pos + offsets[i]

    ax[0].bar(
        xpos,
        means,
        width=width,
        color=palette[i],
        alpha=1.0,
        linewidth=0.0,
        edgecolor='none'
    )

    ax[0].errorbar(
        xpos,
        means,
        yerr=stds,
        fmt='none',
        ecolor='black',
        elinewidth=0.5,
        capsize=2,
        capthick=0.5,
        zorder=2
    )

# -------------------------
# DOTPLOT (matplotlib scatter)
# -------------------------
for i, tgt in enumerate(hue_order):

    sub_raw = data[data['target'] == tgt]

    x = np.array([
        np.where(order == eid)[0][0]
        for eid in sub_raw['eid']
    ])

    x = x + offsets[i]

    ax[0].scatter(
        x,
        sub_raw['auc'].values,
        s=6,
        alpha=1.0,
        color=palette[i],
        edgecolor="white",
        linewidth=0.25,
        zorder=3
    )

ax[0].set_xticks(x_pos)
ax[0].set_xticklabels([EIDS_TO_CELLTYPES[eid] for eid in order])

ax[0].set_ylabel("")
ax[0].set_xlabel("")
ax[0].set_ylim(0.55, 1.0)


# =========================================================
# RIGHT PANEL (DeepTX)
# =========================================================
data = pd.read_csv('extra/results/bs_bf_deeptx_indicators.csv')
data['auc'] = data['auc'] / 100

summary = data.groupby(['eid', 'target'])['auc'].agg(['mean', 'std']).reset_index()

# -------------------------
# BAR + ERRORBAR
# -------------------------
for i, tgt in enumerate(hue_order):
    sub = summary[summary['target'] == tgt]

    means = np.array([sub[sub['eid'] == eid]['mean'].values[0] for eid in order])
    stds  = np.array([sub[sub['eid'] == eid]['std'].values[0] for eid in order])

    xpos = x_pos + offsets[i]

    ax[1].bar(
        xpos,
        means,
        width=width,
        color=palette[i],
        alpha=1.0,
        linewidth=0.0,
        edgecolor='none'
    )

    ax[1].errorbar(
        xpos,
        means,
        yerr=stds,
        fmt='none',
        ecolor='black',
        elinewidth=0.5,
        capsize=2,
        capthick=0.5,
        zorder=2
    )

# -------------------------
# DOTPLOT (matplotlib scatter)
# -------------------------
for i, tgt in enumerate(hue_order):

    sub_raw = data[data['target'] == tgt]

    x = np.array([
        np.where(order == eid)[0][0]
        for eid in sub_raw['eid']
    ])

    x = x + offsets[i]

    ax[1].scatter(
        x,
        sub_raw['auc'].values,
        s=6,
        alpha=1.0,
        color=palette[i],
        edgecolor="white",
        linewidth=0.25,
        zorder=3
    )

ax[1].set_xticks(x_pos)
ax[1].set_xticklabels([EIDS_TO_CELLTYPES[eid] for eid in order])

ax[1].set_ylabel("")
ax[1].set_xlabel("")
ax[1].set_ylim(0.55, 1.0)


# =========================================================
# LEGEND
# =========================================================
legend_patches = [
    mpatches.Patch(color=palette[0], label='BF'),
    mpatches.Patch(color=palette[1], label='BS'),
]

fig.legend(handles=legend_patches, loc='upper center', ncol=2, frameon=False).remove()

fig.savefig(
    'extra/figures/v8/Figure2A.pdf',
    format='pdf',
    bbox_inches='tight'
)

plt.show()
```

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import scipy.stats as stats

EIDS_TO_CELLTYPES = {'E003': 'H1', 'E116': 'GM12878', 'E118': 'HepG2'}

order = pd.Index(['E003', 'E118', 'E116'], name='marks')

input_csv = 'extra/results/predictions_bs_bf.csv'
source_dir = 'scripts/figures/v8/source_data'
figure_path = 'extra/figures/v8/Figure2C.eps'

os.makedirs(source_dir, exist_ok=True)
os.makedirs(os.path.dirname(figure_path), exist_ok=True)

# =========================================================
# LOAD DATA
# =========================================================
data = pd.read_csv(input_csv)

# =========================================================
# SOURCE DATA
# =========================================================
source_data = data.copy()
source_data['cell_type'] = source_data['eid'].map(EIDS_TO_CELLTYPES)

source_columns = ['gene_id', 'eid', 'cell_type', 'fold', 'bf_label_label', 'bf_label_score', 'bs_label_label', 'bs_label_score']
source_data = source_data[[col for col in source_columns if col in source_data.columns]]

source_data.to_csv(f'{source_dir}/fig2c.csv', index=False)
print(f'Source data saved to {source_dir}/fig2c.csv')

# =========================================================
# EXACT N FOR FIG. 2C
# statistical unit = unique gene
# =========================================================
bf_n = data.groupby(['eid', 'bf_label_label'])['gene_id'].nunique().reset_index(name='n').rename(columns={'bf_label_label': 'label'})
bf_n['target'] = 'BF'

bs_n = data.groupby(['eid', 'bs_label_label'])['gene_id'].nunique().reset_index(name='n').rename(columns={'bs_label_label': 'label'})
bs_n['target'] = 'BS'

n_summary = pd.concat([bf_n, bs_n], ignore_index=True)
n_summary['cell_type'] = n_summary['eid'].map(EIDS_TO_CELLTYPES)
n_summary['group'] = n_summary['label'].map({0: 'Low', 1: 'High'})

n_summary['eid'] = pd.Categorical(n_summary['eid'], categories=order, ordered=True)
n_summary['target'] = pd.Categorical(n_summary['target'], categories=['BF', 'BS'], ordered=True)
n_summary['label'] = pd.Categorical(n_summary['label'], categories=[0, 1], ordered=True)

n_summary = n_summary.sort_values(['target', 'eid', 'label']).reset_index(drop=True)

n_summary.to_csv(f'{source_dir}/fig2c_n.csv', index=False)
print(f'Exact n table saved to {source_dir}/fig2c_n.csv')

print('\nExact n for Fig. 2c:')
print(n_summary[['target', 'eid', 'cell_type', 'group', 'n']].to_string(index=False))

# =========================================================
# LEGEND-READY EXACT N
# =========================================================
print('\nLegend-ready exact n values:')

for target in ['BF', 'BS']:
    print(f'\n{target}:')
    sub = n_summary[n_summary['target'] == target]

    for eid in order:
        tmp = sub[sub['eid'] == eid]
        n_low = int(tmp[tmp['group'] == 'Low']['n'].iloc[0])
        n_high = int(tmp[tmp['group'] == 'High']['n'].iloc[0])
        print(f'{EIDS_TO_CELLTYPES[eid]}: n={n_low} low-{target} genes, n={n_high} high-{target} genes')

# =========================================================
# LEGEND TEXT
# =========================================================
bf_parts = []
bs_parts = []

for eid in order:
    cell_type = EIDS_TO_CELLTYPES[eid]

    tmp_bf = n_summary[(n_summary['target'] == 'BF') & (n_summary['eid'] == eid)]
    bf_low = int(tmp_bf[tmp_bf['group'] == 'Low']['n'].iloc[0])
    bf_high = int(tmp_bf[tmp_bf['group'] == 'High']['n'].iloc[0])
    bf_parts.append(f'{cell_type}: n={bf_low} BF− and n={bf_high} BF+ genes')

    tmp_bs = n_summary[(n_summary['target'] == 'BS') & (n_summary['eid'] == eid)]
    bs_low = int(tmp_bs[tmp_bs['group'] == 'Low']['n'].iloc[0])
    bs_high = int(tmp_bs[tmp_bs['group'] == 'High']['n'].iloc[0])
    bs_parts.append(f'{cell_type}: n={bs_low} BS− and n={bs_high} BS+ genes')

print('\nSuggested legend text:')
print('For BF, ' + '; '.join(bf_parts) + '.')
print('For BS, ' + '; '.join(bs_parts) + '.')

# =========================================================
# PLOT
# =========================================================
palette1 = ['#F1C3C8', '#E94C72']
palette2 = ['#B2C1E3', '#036EB8']

fig, ax = plt.subplots(1, 2, constrained_layout=True, figsize=(7.5 / 2.54, 3.5 / 2.54))

sns.boxplot(data=data, x='eid', y='bf_label_score', hue='bf_label_label', fliersize=0, ax=ax[0], palette=palette1, linewidth=0.3, order=order, width=0.5, gap=0.2, saturation=1)

ax[0].set_ylabel('BF score', fontsize=6)
ax[0].set_xlabel('', fontsize=6)
ax[0].set_xticklabels([EIDS_TO_CELLTYPES[eid] for eid in order])
ax[0].tick_params(axis='x', labelsize=5.5, width=0.5)
ax[0].tick_params(axis='y', labelsize=5.5, width=0.5)
ax[0].legend().remove()

sns.boxplot(data=data, x='eid', y='bs_label_score', hue='bs_label_label', fliersize=0, ax=ax[1], palette=palette2, linewidth=0.3, order=order, width=0.5, gap=0.2, saturation=1)

ax[1].set_ylabel('BS score', fontsize=6)
ax[1].set_xlabel('', fontsize=6)
ax[1].set_xticklabels([EIDS_TO_CELLTYPES[eid] for eid in order])
ax[1].tick_params(axis='x', labelsize=5.5, width=0.5)
ax[1].tick_params(axis='y', labelsize=5.5, width=0.5)
ax[1].legend().remove()

legend_patches = [
    mpatches.Patch(color=palette1[1], label='High burst freq. genes'),
    mpatches.Patch(color=palette1[0], label='Low burst freq. genes'),
    mpatches.Patch(color=palette2[1], label='High burst size genes'),
    mpatches.Patch(color=palette2[0], label='Low burst size genes'),
]

plt.savefig(figure_path, format='eps', bbox_inches='tight')
print(f'\nFigure saved to {figure_path}')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.colors as mcolors

# =========================
# Load data
# =========================
data = pd.read_csv("extra/results/cross_cell_type_bs_bf_para.csv", sep="\t")

# Avoid repeated division if values are already 0-1
if data["bf_auc"].max() > 1:
    data["bf_auc"] = data["bf_auc"] / 100
if data["bs_auc"].max() > 1:
    data["bs_auc"] = data["bs_auc"] / 100

order = ["E003", "E118", "E116"]

eid_labels = {
    "E003": "H1",
    "E118": "HepG2",
    "E116": "GM12878"
}

# =========================
# Build matrices
# rows    = training cell type
# columns = validation cell type
# values  = median AUROC across folds
# =========================
bf_mat = (
    data
    .pivot_table(
        index="train_eid",
        columns="valid_eid",
        values="bf_auc",
        aggfunc="mean"
    )
    .loc[order, order]
)

bs_mat = (
    data
    .pivot_table(
        index="train_eid",
        columns="valid_eid",
        values="bs_auc",
        aggfunc="mean"
    )
    .loc[order, order]
)

# Display labels
bf_plot = bf_mat.rename(index=eid_labels, columns=eid_labels)
bs_plot = bs_mat.rename(index=eid_labels, columns=eid_labels)

print("BF AUROC median matrix:")
print(bf_plot)

print("\nBS AUROC median matrix:")
print(bs_plot)

# =========================
# Unified colormap
# =========================
COLORS = ['#FFFDDF', "#7FCDBB", "#225EA8"]
cmap = mcolors.LinearSegmentedColormap.from_list(
    "colormap",
    COLORS,
    N=100
)

vmin = 0.74
vmax = 1.00

# =========================
# Plot
# =========================
fig, ax = plt.subplots(
    1, 2,
    figsize=(8.5 / 2.54, 3.2 / 2.54),
    constrained_layout=True
)

sns.heatmap(
    bf_plot,
    ax=ax[0],
    cmap=cmap,
    vmin=vmin,
    vmax=vmax,
    annot=True,
    fmt=".2f",
    linewidths=0.3,
    linecolor="white",
    cbar_kws={
        "label": "BF AUROC",
        "shrink": 0.6,
        "fraction": 0.05
    },
    annot_kws={
        "fontsize": 5
    }
)

ax[0].set_title("Burst frequency", fontsize=6)
ax[0].set_xlabel("Validation cell type", fontsize=6)
ax[0].set_ylabel("Training cell type", fontsize=6)
ax[0].tick_params(axis="both", labelsize=5, length=0)

sns.heatmap(
    bs_plot,
    ax=ax[1],
    cmap=cmap,
    vmin=vmin,
    vmax=vmax,
    annot=True,
    fmt=".2f",
    linewidths=0.3,
    linecolor="white",
    cbar_kws={
        "label": "BS AUROC",
        "shrink": 0.6,
        "fraction": 0.05
    },
    annot_kws={
        "fontsize": 5
    }
)

ax[1].set_title("Burst size", fontsize=6)
ax[1].set_xlabel("Validation cell type", fontsize=6)
ax[1].set_ylabel("", fontsize=6)
ax[1].tick_params(axis="both", labelsize=5, length=0)

# =========================
# Style cleanup
# =========================
for a in ax:
    a.set_xticklabels(a.get_xticklabels(), rotation=0)
    a.set_yticklabels(a.get_yticklabels(), rotation=0)

    cbar = a.collections[0].colorbar
    cbar.ax.tick_params(labelsize=5, length=2, width=0.3)
    cbar.set_label(cbar.ax.get_ylabel(), fontsize=6)

plt.savefig(
    "extra/figures/v8/Figure2G_heatmap.pdf",
    format="pdf",
    bbox_inches="tight"
)

plt.show()
```

```python

eid = 'E116'

def compute_label_statistics(eid,threshold=0):

    labels_deeptx = pd.read_csv(f'extra/datasets/burst/labels/deeptx/{EIDS_TO_CELLTYPES[eid]}_estimated_burst.csv')
    labels_deeptx = labels_deeptx.fillna(0.0)
    bs_median = np.median(labels_deeptx['bs'])
    bf_median = np.median(labels_deeptx['bf'])

    labels_deeptx['bs_label'] = labels_deeptx.apply(lambda row: 1 if row['bs'] > bs_median else 0, axis=1)
    labels_deeptx['bf_label'] = labels_deeptx.apply(lambda row: 1 if row['bf'] > bf_median else 0, axis=1)

    labels_txburst = pd.read_csv(f'extra/datasets/burst/processed/{EIDS_TO_CELLTYPES[eid]}_statistic_gene_transcript_region_delay_1.csv')
    bs_median = np.median(labels_txburst['bs'])
    bf_median = np.median(labels_txburst['bf'])
    labels_txburst['mean'] = labels_txburst['bf']*labels_txburst['bs']

    labels_txburst['bs_label'] = labels_txburst.apply(lambda row: 1 if row['bs'] > bs_median else 0, axis=1)
    labels_txburst['bf_label'] = labels_txburst.apply(lambda row: 1 if row['bf'] > bf_median else 0, axis=1)

    data = pd.merge(labels_deeptx[['gene_name','bf','bs',f'bf_label','bs_label','mean_true']],labels_txburst[['gene_name','bf','bs','bf_label','bs_label','mean']],on='gene_name',suffixes=['_deeptx','_txburst'])
    data_exp = data[data['mean'] > threshold]
    data_no_exp = data[data['mean'] <= threshold]
    # data.to_csv('tests/test.csv',index=False)
    consistency = {
        'eid':eid,
        'bf':[(data_exp[data_exp[f'bf_label_deeptx']==data_exp[f'bf_label_txburst']].shape[0] + data_no_exp.shape[0])/data.shape[0]],
        'bs':[(data_exp[data_exp[f'bs_label_deeptx']==data_exp[f'bs_label_txburst']].shape[0] + data_no_exp.shape[0])/data.shape[0]]
        }

    return consistency

eid = 'E003'
threshold = 0
consistency_E003 = compute_label_statistics(eid,threshold)
consistency_E003 = pd.DataFrame(consistency_E003)

eid = 'E118'
consistency_E116 = compute_label_statistics(eid,threshold)
consistency_E116 = pd.DataFrame(consistency_E116)

eid = 'E116'
consistency_E118 = compute_label_statistics(eid,threshold)
consistency_E118 = pd.DataFrame(consistency_E118)


consistency_df = pd.concat([consistency_E003, consistency_E116, consistency_E118], axis=0)
print(consistency_df)

consistency_df = pd.melt(
    consistency_df,
    id_vars='eid',             # 不展开 eid 列
    value_vars=['bf', 'bs'],   # 要展开的列
    var_name='burst_type',     # 新列名：表示是 bf 还是 bs
    value_name='consistency'   # 新列名：对应的值
)


order = pd.Index(["E003","E118","E116"],name='marks')
data = pd.read_csv(f'extra/results/predictions_bs_bf.csv')

fig, ax = plt.subplots(1, 1, constrained_layout=True, figsize=(3.5/2.54, 3.5/2.54))
palette1 = ["#DD2421","#036EB8"]
sns.barplot(
    data=consistency_df,
    x='eid',
    y='consistency',
    hue='burst_type',
    palette=palette1,
    linewidth=0.5,
    saturation=1,
    order=order,
)

ax.set_ylabel('Label consistency between\nDeepTx and txburst', fontsize=6)
ax.set_xlabel('',fontsize=6)
ax.set_xticklabels([EIDS_TO_CELLTYPES[eid] for eid in order],rotation=0,fontsize=5)
ax.set_ylim(0.5,1.0)
ax.legend().remove()

import matplotlib.patches as mpatches
legend_patches = [
    mpatches.Patch(color=palette1[0], label='BF'),
    mpatches.Patch(color=palette1[1], label='BS'),
]
fig.legend(
    handles=legend_patches, 
    title="", 
    handlelength=2/2.54,      # 控制色块的水平长度
    handleheight=1.7/2.54,      # 控制色块的垂直高度（仅在 fancybox=True 时有明显效果）
    loc="upper left", 
    bbox_to_anchor=(0.98, 1.02), 
    title_fontsize=6,
    fontsize=5, 
    ncol=1,
    edgecolor='none')

plt.show()
```

```python
def compute_sampling_probabillity_consistency(data1, data2,threshold=0.0):


    data = pd.merge(data1, data2, on='gene_id', suffixes=('_delay1', '_delay2'))
    data_exp = data[data['mean_delay2'] >= threshold]
    data_no_exp = data[data['mean_delay2'] < threshold]

    consistency = {
        'eid':eid,
        'bf':[(data_exp[data_exp[f'bf_label_delay1']==data_exp[f'bf_label_delay2']].shape[0] + data_no_exp.shape[0])/data.shape[0]],
        'bs':[(data_exp[data_exp[f'bs_label_delay1']==data_exp[f'bs_label_delay2']].shape[0] + data_no_exp.shape[0])/data.shape[0]]
        }

    return consistency
```
