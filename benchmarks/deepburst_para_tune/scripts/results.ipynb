## binsize tuning

```python
targets=['bs_label','bf_label']
with open('benchmarks/deepburst_para_tune/results/bs_bf_para_tune_binsize_indicators.csv','w') as w:
    line = ['eid','fold','auc','target','binsize']
    w.write(','.join(line) + '\n')
    for eid in ["E003","E116","E118"]:            
        for fold in [0, 1, 2, 3]:
            for binsize in ['100','200','500', '1000', '2000']:
                # checkpoints = f"benchmarks/deepburst_para_tune/checkpoints/{eid}.{fold}.bs_bf_para.tune.binsize_{binsize}.model.pt"
                checkpoints = f"checkpoints/{eid}.{fold}.binsize_{binsize}.bs_bf_para.model.pt"
                ckpt = torch.load(checkpoints,map_location=DEVICE)
                for target in targets:
                    auc = ckpt[f"{target}_val_auc"]
                    line = [eid,str(fold),str(auc),target, binsize]
                    w.write(','.join(line) + '\n')

```

```python
data = pd.read_csv('benchmarks/deepburst_para_tune/results/bs_bf_para_tune_binsize_indicators.csv')
data['auc'] = data['auc'] / 100

order = ['E003','E118','E116']
fig, ax = plt.subplots(1, 2, figsize=(10.3/2.54, 3.75/2.54))
palette = ['#F0D5D3','#E8C1BD','#D79792','#D0837E','#BF392F']


sns.boxplot(data=data[data['target']=='bf_label'],y='auc',x='eid',hue='binsize',fliersize=0, linewidth=0.2,order=order,
            gap=0.2,width=0.8,palette=palette,ax=ax[0],saturation=1)


ax[0].set_ylim(bottom=0.85,top=0.95)
ax[0].set_xlabel('')
ax[0].set_ylabel(f'BF AUROC\n(4-fold cross-validation)',fontsize=6)
# ax[0].set_ylabel(f'',fontsize=6)
ax[0].legend().remove()
ax[0].set_xticklabels([EIDS_TO_CELLTYPES[eid] for eid in order], fontsize=6)
ax[0].set_yticks([i/100 for i in range(85, 96, 2)],labels=[i/100 for i in range(85, 96, 2)])


palette = ['#D5E2F0','#BFD3E8','#95B6D9','#81A8D1','#568AC2']
sns.boxplot(data=data[data['target']=='bs_label'],y='auc',x='eid',hue='binsize', fliersize=0, linewidth=0.2,order=order,
            gap=0.2,width=0.8,palette=palette,ax=ax[1],saturation=1)

ax[1].set_ylabel(f'BS AUROC\n(4-fold cross-validation)',fontsize=6)
# ax[1].set_ylabel(f'',fontsize=6)
ax[1].set_ylim(bottom=0.79,top=0.94)
ax[1].set_xlabel('')
ax[1].legend().remove()
ax[1].set_xticklabels([EIDS_TO_CELLTYPES[eid] for eid in order], fontsize=6)
ax[1].set_yticks([i/100 for i in range(79, 95, 5)],labels=[i/100 for i in range(79, 95, 5)])

plt.tight_layout()
plt.savefig(f'benchmarks/deepburst_para_tune/figures/v7/Supplementary_Figure7C.eps', format='eps', bbox_inches='tight')
plt.show()
```
