```python
import numpy as np
import pandas as pd 

TARGET_PHASE = 'G2M'
PHASE_ORIGIN = 'deepcycle'
delay = 0.5

inferred_results = pd.read_csv(f"extra/datasets/burst/cellcycle_deeptx/gm12878/inferred_results_{TARGET_PHASE}_{PHASE_ORIGIN}_delay_{delay}.csv")
infer_columns = ['x1', 'x2', 'x3', 'x4', 'k_syn','gene_name','bf','bs', 'mean_es', 'mean_true','var_true']


meta_data  = pd.read_csv("extra/datasets/processed/v2/meta_datasets/meta_data_E116_deeptx.csv")

meta_data_G1 = meta_data.drop(columns=['cv','mean','cv_label','mean_label','bs','bf','bs_label','bf_label'])
meta_data_G1


meta_data_G1 = pd.merge(meta_data_G1, inferred_results[['gene_name', 'bs', 'bf']], on='gene_name', how='inner')
meta_data_G1


bs_median = np.median(meta_data_G1['bs'])
bf_median = np.median(meta_data_G1['bf'])



meta_data_G1['bs_label'] = meta_data_G1.apply(lambda row: 1 if row['bs'] > bs_median else 0, axis=1)
meta_data_G1['bf_label'] = meta_data_G1.apply(lambda row: 1 if row['bf'] > bf_median else 0, axis=1)
meta_data_G1.to_csv(f"extra/datasets/processed/v2/meta_datasets/meta_data_E116_deeptx_delay_{delay}_with_cellsize_1_cellcycle_{TARGET_PHASE}_{PHASE_ORIGIN}.csv", index=False)
```
